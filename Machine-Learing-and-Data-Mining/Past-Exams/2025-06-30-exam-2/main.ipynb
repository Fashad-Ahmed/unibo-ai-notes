```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

target = "y"

X = df.drop(columns=[target])
y = df[target]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.7, random_state=42)
```

```python
lr = LinearRegression()
lr.fit(X_train, y_train)
predict = lr.predict(X_test)

predict
```

```python
from sklearn.metrics import mean_squared_error

rmse = np.sqrt(mean_squared_error(predict, y_test))

print(rmse)
```

```python
X_corr = df.drop(columns=["X03", "X04", "X07"])

X_train, X_test, y_train, y_test = train_test_split(X_corr, y, train_size=0.7, random_state=42)


lr = LinearRegression()
lr.fit(X_train, y_train)
predict = lr.predict(X_test)

predict
```

```python
from sklearn.tree import DecisionTreeRegressor

dtr = DecisionTreeRegressor(random_state=42)


dtr.fit(X_train, y_train)
predict = dtr.predict(X_test)

rmse = np.sqrt(mean_squared_error(predict, y_test))

print(rmse)
```

```python
max_depth = dtr.tree_.max_depth
max_depth
```

```python
dtr.tree_.impurity[0]
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import make_scorer

parameter_values = range(1, max_depth + 1)

tune_param_dt = [{
    "max_depth": [*parameter_values],
    "criterion": ["squared_error"],    
}]

rmse_scorer = make_scorer(
    lambda y_true, y_pred: np.sqrt(mean_squared_error(y_true, y_pred)),
    greater_is_better=False
)


rmse_scorer
```

```python
clf = GridSearchCV(dtr, tune_param_dt, scoring=rmse_scorer, cv=5)
```

```python
import numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error, make_scorer

# Features and target
X = df.drop(columns='y')
y = df['y']

# Reduced feature set (drop loosely correlated columns)
X_reduced = X.drop(columns=['X00', 'X07'])

# Split off a hold-out test set (25% of the data)
X_train, X_test, y_train, y_test = train_test_split(
    X_reduced, y, test_size=0.25, random_state=42
)

# RMSE scorer (negative because GridSearchCV maximizes the score)
rmse_scorer = make_scorer(
    lambda y_true, y_pred: np.sqrt(mean_squared_error(y_true, y_pred)),
    greater_is_better=False
)

# Decision Tree model
dt = DecisionTreeRegressor(random_state=42)

tune_param_dt = [{
    "max_depth": [*parameter_values],
    "criterion": ["squared_error"],    
}]

# Grid search with 5-fold cross-validation on training set
grid = GridSearchCV(
    dt,
    tune_param_dt,
    scoring=rmse_scorer,
    cv=5
)

# Fit GridSearchCV
grid.fit(X_train, y_train)

# Best model from CV
best_model = grid.best_estimator_

# Evaluate final model on hold-out test set
y_pred_test = best_model.predict(X_test)
final_rmse = np.sqrt(mean_squared_error(y_test, y_pred_test))

final_rmse
```
